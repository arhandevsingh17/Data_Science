```python
z = input()
pal = True
i = z[0]
j = z[len(z)-1]
while(i<j):
  if(i!=j):
    pal = False
    break
    i+=1
    j-=1
if(i==j):
  print(pal)
else:
  print("false")
```

```python
#sort
list2 = [1,3,5,7,5,67,48,6,5,]
list2.sort()
print(list2)
#reverse
list2.reverse()
print(list2)
#sort Descending
list2.sort(reverse=True)
print(list2)
#sum
print(sum(list2))
#min
print(min(list2))
#max
print(max(list2))
#count
print(list2.count(5))
#sorted
print(sorted(list2)) #creates a new sorted list
print("///")
#copy
list3 = list2.copy()
print(list3)
#clear
list2.clear()
print(list2)
```
